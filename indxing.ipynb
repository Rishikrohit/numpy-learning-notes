***numpy indexing and slicing of matrix***

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[1]
```

```python
arr[0:4]
```

```python
print(arr[0,0])
```

```python
arr[0,2]
```

slicing


```python
slice = arr[0:2,1:3]
slice
```
